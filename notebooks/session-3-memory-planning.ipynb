```python
from groq import Groq
from getpass import getpass

api_key = getpass("Enter your Groq API key: ")
client = Groq(api_key=api_key)
```

```python
def chat(message):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": message
            }
        ]
    )
    return response.choices[0].message.content


print(chat("Hi! My name is Aarav."))
print(chat("What is my name?"))
```

```python
memory = []

def chat_with_memory(message):

    memory.append({
        "role": "user",
        "content": message
    })

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=memory
    )

    answer = response.choices[0].message.content

    memory.append({
        "role": "assistant",
        "content": answer
    })

    return answer


print(chat_with_memory("Hi, my name is Aarav"))

print(chat_with_memory("What is my name?"))
```

```python
def make_plan(task):
    prompt = f"""
    Break the task into 4 clear, sequential steps.

    TASK:
    {task}

    Return only numbered steps.
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


task = "Prepare a seminar on Artificial Intelligence"

plan = make_plan(task)

print(plan)
```

```python
memory = []

steps = [
    "Choose three beginner-level topics for a Machine Learning workshop.",
    "Create a 60-minute schedule using those three topics.",
    "Create one hands-on activity for each topic.",
    "Create the final workshop outline using everything produced so far."
]

def run_step(step, memory):

    prompt = f"""
    We are working on this task:

    Create a beginner-level Machine Learning workshop.

    Previous work stored in memory:
    {memory}

    Current step:
    {step}

    Complete ONLY the current step.

    Use the previous work as context.
    Do not repeat previous work unnecessarily.
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    result = response.choices[0].message.content

    memory.append({
        "step": step,
        "result": result
    })

    return result


for i, step in enumerate(steps, 1):

    print(f"\n{'='*50}")
    print(f"STEP {i}")
    print(f"{'='*50}")

    result = run_step(step, memory)

    print(result)

    input("\nThen press ENTER to continue...")


print("\n\n" + "="*50)
print("FINAL MEMORY")
print("="*50)

for item in memory:

    print("\nSTEP:")
    print(item["step"])

    print("\nRESULT:")
    print(item["result"])
```
